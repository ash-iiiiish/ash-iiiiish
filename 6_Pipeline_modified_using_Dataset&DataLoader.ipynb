# Dataset Loading

```python
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
df.head()
```

```python
df.shape
```

```python
df.drop(columns = ['id','Unnamed: 32'],inplace =True)
```

```python
df.shape
```

```python
df['diagnosis'].value_counts()
```

# Train Test Split

```python
X_train,X_test,y_train,y_test = train_test_split(df.iloc[:,1:],df.iloc[:,0],test_size = 0.3)
```

```python
print(f"Shape of X_train : {X_train.shape}")
print(f"Shape of X_test : {X_test.shape}")
print(f"Shape of y_train : {y_train.shape}")
print(f"Shape of y_test : {y_test.shape}")
```

# Scaling

```python
# scaling is used to send similar type of data to our model
scaler = StandardScaler()
X_train =scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train
```

```python
X_test
```

# Label Encoding


```python
y_train #our diagnosis is in alphabet format i.e. B and M which is not understandable by our machine so we have to encode these alphabets
```

```python
encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)
```

```python
y_train # now 0 : B and 1 : M
```

# Numpy arrays to Pytorch Tensors

```python
X_train_tensor = torch.FloatTensor(X_train)
X_test_tensor = torch.FloatTensor(X_test)
y_train_tensor = torch.LongTensor(y_train)
y_test_tensor = torch.LongTensor(y_test)
```

```python
print(f"shape of X_train_tensor : {X_train_tensor.shape}")
print(f"shape of X_test_tensor : {X_test_tensor.shape}")
print(f"shape of y_train_tensor : {y_train_tensor.shape}")
print(f"shape of y_test_tensor : {y_test_tensor.shape}")
```

```python
from torch.utils.data import Dataset,DataLoader

class CustomDataset(Dataset):

  def __init__(self,features,labels):
    self.features=features
    self.labels=labels


  def __len__(self):
    return len(self.features)

  def __getitem__(self,idx):
    return self.features[idx],self.labels[idx]
```

```python
train_dataset = CustomDataset(X_train_tensor,y_train_tensor)
test_dataset = CustomDataset(X_test_tensor,y_test_tensor)
```

```python
train_dataset[10]
```

```python
test_dataset[10]
```

```python
train_loader = DataLoader(train_dataset,batch_size=32,shuffle=True)
test_loader = DataLoader(test_dataset,batch_size=32,shuffle=True)
```

```python
import torch
import torch.nn as nn

class MySimpleNN(torch.nn.Module):
  def __init__(self,num_features):
    super().__init__()
    self.network=nn.Sequential(
        nn.Linear(num_features,1),
        nn.Sigmoid()
    )

  def forward(self,features):
    out=self.network(features)
    return out
```

# Important Parameters

```python
learning_rate = 0.1
epochs = 30
```

```python
#define loss function
loss_function = nn.BCELoss()
```

# Training Pipeline

```python
# create model
# model = MySimpleNN(X_train_tensor)
# model.weights #matrix of 30 by 1
```

```python
# model.bias# a scalar with item 0
```

```python
#define model
model = MySimpleNN(X_train_tensor.shape[1])
#define loop

#define optimizer
optimizer = torch.optim.SGD(model.parameters(),lr=learning_rate)

for epoch in range(epochs):
  for batch_features,batch_labels in train_loader:

    #forward pass
    y_pred=model(batch_features)

    #loss calculate
    loss=loss_function(y_pred,batch_labels.view(-1,1).float())

    #clear gradients
    optimizer.zero_grad()

    #backward pass
    loss.backward()

    #parameters update
    optimizer.step()

  print(f"Epoch : {epoch+1}, Loss : {loss.item()}")
```

# Evaluation

```python
# Model evaluation using test_loader
model.eval() # set the model to evaluation mode
accuracy_list = []

with torch.no_grad():
  for batch_features,batch_labels in test_loader:

    #forward pass
    y_pred=model(batch_features)
    y_pred=(y_pred>0.8).float() #convert probabilities to binary predictions

    #calculate accuracy for the current batch
    batch_accuracy=(y_pred.view(-1)==batch_labels).float().mean().item()
    accuracy_list.append(batch_accuracy)

# calculate overall accuracy
overall_accuracy=sum(accuracy_list)/len(accuracy_list)
print(f"Accuracy : {overall_accuracy:.4f}")
```

# Defining the Model

```python
import torch.nn as nn

class MyModel(nn.Module):

  def __init__(self,num_features):

    super().__init__()
    self.network=nn.Sequential(
        nn.Linear(num_features,1),
        nn.Sigmoid()
    )

  def forward(self,features):
    out = self.network(features)
    return out
```

# Important Parameters

```python
learning_rate = 0.1
epochs = 25
```

```python
#define loss function
loss_function = nn.BCELoss()
```

```python
#define optimizer
# Instantiate the new model
model = MyModel(X_train_tensor.shape[1])
optimizer=torch.optim.SGD(model.parameters(),lr=learning_rate)
```

# Training Pipeline

```python
# create model
model =MyModel(X_train_tensor.shape[1])

#define loop for epoch times :
for epoch in range(epochs):

  #forward pass
  y_pred=model(X_train_tensor)

  #loss calculate
  loss=loss_function(y_pred,y_train_tensor.reshape(-1,1).float())

  #backward pass
  loss.backward()

  #parameters update
  optimizer.step()

  #zero gradients
  optimizer.zero_grad()

  #print loss in each epochs
  print(f"Epoch : {epoch+1}, Loss : {loss.item()}")
```

# Evaluation

```python
# model evaluation
with torch.no_grad():
  y_pred=model.forward(X_test_tensor)
  y_pred =(y_pred > 0.9).float()
  accuracy = (y_pred==y_test_tensor).float().mean()
  print(f"Accuracy : {accuracy.item():.4f}")
```
